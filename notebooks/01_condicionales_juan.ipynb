```python
import csv

# Lectura del archivo CSV desde la carpeta data/
archivo = open("../data/calidad_agua_cundinamarca_gov.csv", mode="r", encoding="utf-8")
lector = list(csv.DictReader(archivo))
archivo.close()

# Seleccionamos el registro 4 (Soacha - Quebrada Tibaníes) de forma directa
registro = lector[4]

municipio = registro["municipio"]
divipola = registro["codigo_divipola"]
ph = float(registro["ph"])
turbiedad = float(registro["turbiedad_unt"])
cloro = float(registro["cloro_residual_mgL"])
irca = float(registro["irca_porcentaje"])

print(f"=== DIAGNÓSTICO NORMATIVO (DANE {divipola} - {municipio}) ===")
```

```python
# 1. Clasificación Nivel de Riesgo IRCA (Resolución 2115 de 2007)
if irca <= 5.0:
    nivel_riesgo = "SIN RIESGO (Agua apta para consumo humano)"
elif irca <= 14.0:
    nivel_riesgo = "RIESGO BAJO (Susceptible de mejoramiento)"
elif irca <= 35.0:
    nivel_riesgo = "RIESGO MEDIO (No apta - Gestión directa)"
elif irca <= 80.0:
    nivel_riesgo = "RIESGO ALTO (No apta - Alerta sanitaria)"
else:
    nivel_riesgo = "INVIABLE SANITARIAMENTE"

# 2. Validación de Cloro Residual Libre (Rango: 0.3 a 2.0 mg/L)
if cloro < 0.3:
    estado_cloro = "Deficiente (Riesgo microbiológico)"
elif cloro > 2.0:
    estado_cloro = "Exceso (Riesgo toxicológico)"
else:
    estado_cloro = "Cumple norma"

# 3. Validación de pH (Rango: 6.5 a 9.0)
if 6.5 <= ph <= 9.0:
    estado_ph = "Dentro de norma"
else:
    estado_ph = "Fuera de rango permisible"

print(f"Punto de muestreo: {registro['punto_muestreo']}")
print(f"IRCA ({irca}%): {nivel_riesgo}")
print(f"Cloro residual ({cloro} mg/L): {estado_cloro}")
print(f"pH ({ph}): {estado_ph}")
```
