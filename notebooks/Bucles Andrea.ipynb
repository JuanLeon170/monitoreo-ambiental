**Fuente:** Datos Abiertos Colombia (datos.gov.co) / Codificación DANE DIVIPOLA  
**Nota metodológica:** En este notebook usamos contadores, saltos en rangos y sentencias de parada (*break*).

```python
import csv

# Para abrir el archivo
archivo = open("calidad_agua_cundinamarca_gov.csv", encoding="utf-8")
registros = list(csv.DictReader(archivo))
archivo.close()


# Se recorren los registros con un bucle FOR y se inicia a CONTAR desde cada estación analizada
contador = 0
for fila in registros:
    contador += 1  # Suma 1 en cada vuelta
    print(f"Estación #{contador}: Municipio {fila['municipio']} -> IRCA: {fila['irca_porcentaje']}%")


# range(inicio, fin, salto) lo puse de dos en dos por fila
print("\nRevisando estaciones de 2 en 2:")
for i in range(0, len(registros), 2):
    fila = registros[i]
    print(f"Posición {i} -> Municipio: {fila['municipio']}")


# Bucle While con 'municipios' totales de la lista, celda irca y se convierten a decimal
print("\nIniciando monitoreo con WHILE:")
indice = 0
while indice < len(registros):
    fila = registros[indice]
    irca = float(fila["irca_porcentaje"])

    # Break
    if irca > 15.0:
        print(f"Límite superado: {fila['municipio']} ({irca}%). Bucle detenido por seguridad.")
        break  # Rompe y apaga el bucle de inmediato

    # Para que el contador avance a la siguiente fila del archivo
    print(f"Analizando {fila['municipio']}: IRCA seguro ({irca}%)...")
    indice += 1
```
